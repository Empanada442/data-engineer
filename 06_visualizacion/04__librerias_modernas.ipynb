# 1. Polars - DataFrame de Alto Performance

## ¿Qué es?
Polars es una alternativa moderna a Pandas orientada a alto rendimiento.

## ¿Para qué sirve?
Acelera operaciones con grandes datasets y mejora la eficiencia en pipelines analíticos.

## ¿Cómo se usa?
- Cargar datos en un DataFrame de Polars  
- Aplicar transformaciones usando expresiones  
- Comparar tiempos de ejecución y uso de memoria frente a Pandas  

````python
print("=" * 70)
print("POLARS: DATAFRAME MODERNO".center(70))
print("=" * 70)

print("""
🐻‍❄️ ¿QUÉ ES POLARS?

Librería de análisis de datos moderna escrita en Rust,
con bindings para Python. Diseñada para ser:
  • RÁPIDA (5-10x más que Pandas)
  • EFICIENTE en memoria
  • EXPRESIVA (API intuitiva)
  • PARALELA por defecto

CARACTERÍSTICAS PRINCIPALES:

1. LENGUAJE EXPRESIVO:
   • Sintaxis similar a dplyr (R) y SQL
   • Chain de operaciones legible
   • Type-safe

2. LAZY EVALUATION:
   • Optimiza queries antes de ejecutar
   • Elimina operaciones redundantes
   • Paraleliza automáticamente

3. PERFORMANCE:
   • Multi-threading sin GIL
   • Vectorización SIMD
   • Zero-copy cuando posible
   • Manejo eficiente de strings

4. FORMATOS:
   • Lee Parquet súper rápido
   • CSV, JSON, Arrow
   • Interop con Pandas

─────────────────────────────────────────────────────────────

🎯 POLARS vs PANDAS

┌─────────────────────────────────────────────────────────┐
│                  POLARS vs PANDAS                       │
├─────────────────────────────────────────────────────────┤
│ VELOCIDAD:                                              │
│   Polars: ⚡⚡⚡⚡⚡ (5-10x más rápido)                   │
│   Pandas: ⚡⚡                                           │
│                                                         │
│ MEMORIA:                                                │
│   Polars: 📊 (Más eficiente)                            │
│   Pandas: 📊📊📊                                        │
│                                                         │
│ API:                                                    │
│   Polars: Expresiones, method chaining                 │
│   Pandas: Más flexible, a veces verbose                │
│                                                         │
│ ECOSISTEMA:                                             │
│   Polars: Creciendo, menos herramientas                │
│   Pandas: Maduro, integración universal                │
│                                                         │
│ LEARNING CURVE:                                         │
│   Polars: Moderada (nuevo paradigma)                   │
│   Pandas: Moderada (conocida por millones)             │
└─────────────────────────────────────────────────────────┘

CUÁNDO USAR POLARS:
  ✓ Datasets grandes (>1GB)
  ✓ Necesitas máxima performance
  ✓ Operaciones complejas de transformación
  ✓ Archivos Parquet grandes
  ✓ Pipelines de datos en producción

CUÁNDO USAR PANDAS:
  ✓ Datasets pequeños (<500MB)
  ✓ Análisis exploratorio rápido
  ✓ Necesitas integración con otras librerías
  ✓ Equipo ya conoce Pandas
  ✓ Prototipado rápido

💡 PUEDES USAR AMBOS:
   Polars para ETL → Convertir a Pandas para análisis/viz
""")

# Intentar importar Polars
print("\n" + "=" * 70)
print("INSTALACIÓN Y CONFIGURACIÓN".center(70))
print("=" * 70)

try:
    import polars as pl
    print(f"✅ Polars {pl.__version__} instalado correctamente")
    polars_available = True
except ImportError:
    print("❌ Polars no está instalado")
    print("\n📦 Para instalar Polars:")
    print("   pip install polars")
    print("   # o")
    print("   conda install -c conda-forge polars")
    polars_available = False

if polars_available:
    print("\n" + "=" * 70)
    print("EJEMPLO 1: SINTAXIS BÁSICA DE POLARS".center(70))
    print("=" * 70)
    
    # Crear DataFrame Polars
    print("\n1️⃣ CREANDO DATAFRAME EN POLARS")
    
    df_polars = pl.DataFrame({
        'id': range(1, 6),
        'nombre': ['Ana', 'Bruno', 'Carlos', 'Diana', 'Elena'],
        'edad': [25, 30, 35, 28, 32],
        'ciudad': ['Madrid', 'Barcelona', 'Madrid', 'Valencia', 'Barcelona'],
        'salario': [35000, 45000, 55000, 40000, 50000]
    })
    
    print(df_polars)
    
    # Información del DataFrame
    print("\n📊 INFORMACIÓN DEL DATAFRAME:")
    print(f"  • Shape: {df_polars.shape}")
    print(f"  • Columnas: {df_polars.columns}")
    print(f"  • Tipos de datos:")
    for col, dtype in zip(df_polars.columns, df_polars.dtypes):
        print(f"    - {col}: {dtype}")
    
    # SELECCIÓN Y FILTRADO
    print("\n" + "=" * 70)
    print("2️⃣ SELECCIÓN Y FILTRADO (Polars Expression API)".center(70))
    print("=" * 70)
    
    # Seleccionar columnas
    print("\n📌 Seleccionar columnas:")
    print(df_polars.select(['nombre', 'edad']))
    
    # Filtrar
    print("\n📌 Filtrar (edad > 30):")
    print(df_polars.filter(pl.col('edad') > 30))
    
    # Combinar operaciones (method chaining)
    print("\n📌 Chain de operaciones:")
    resultado = (
        df_polars
        .filter(pl.col('edad') > 28)
        .select(['nombre', 'ciudad', 'salario'])
        .sort('salario', descending=True)
    )
    print(resultado)
    
    # AGREGACIONES
    print("\n" + "=" * 70)
    print("3️⃣ AGREGACIONES Y GROUP BY".center(70))
    print("=" * 70)
    
    # Agregación simple
    print("\n📊 Estadísticas de salario:")
    stats = df_polars.select([
        pl.col('salario').mean().alias('salario_promedio'),
        pl.col('salario').median().alias('salario_mediana'),
        pl.col('salario').min().alias('salario_minimo'),
        pl.col('salario').max().alias('salario_maximo'),
    ])
    print(stats)
    
    # Group by
    print("\n📊 Salario promedio por ciudad:")
    por_ciudad = (
        df_polars
        .group_by('ciudad')
        .agg([
            pl.col('salario').mean().alias('salario_promedio'),
            pl.col('edad').mean().alias('edad_promedio'),
            pl.count().alias('cantidad')
        ])
        .sort('salario_promedio', descending=True)
    )
    print(por_ciudad)
    
    # TRANSFORMACIONES
    print("\n" + "=" * 70)
    print("4️⃣ TRANSFORMACIONES Y NUEVAS COLUMNAS".center(70))
    print("=" * 70)
    
    df_transformado = df_polars.with_columns([
        (pl.col('salario') * 1.10).alias('salario_con_aumento'),
        (pl.col('salario') / 12).alias('salario_mensual'),
        pl.when(pl.col('edad') < 30)
          .then(pl.lit('Joven'))
          .otherwise(pl.lit('Senior'))
          .alias('categoria_edad')
    ])
    
    print(df_transformado)
    
    # COMPARACIÓN PANDAS vs POLARS
    print("\n" + "=" * 70)
    print("5️⃣ COMPARACIÓN: MISMA OPERACIÓN EN AMBOS".center(70))
    print("=" * 70)
    
    # Crear mismo DataFrame en Pandas
    df_pandas = pd.DataFrame({
        'id': range(1, 6),
        'nombre': ['Ana', 'Bruno', 'Carlos', 'Diana', 'Elena'],
        'edad': [25, 30, 35, 28, 32],
        'ciudad': ['Madrid', 'Barcelona', 'Madrid', 'Valencia', 'Barcelona'],
        'salario': [35000, 45000, 55000, 40000, 50000]
    })
    
    print("\n📝 OPERACIÓN: Personas >28 años, agrupar por ciudad, calcular promedio")
    
    print("\n🐼 PANDAS:")
    print("""
resultado_pandas = (
    df_pandas[df_pandas['edad'] > 28]
    .groupby('ciudad')
    .agg({
        'salario': 'mean',
        'edad': 'mean',
        'id': 'count'
    })
    .rename(columns={'id': 'cantidad'})
    .sort_values('salario', ascending=False)
)
""")
    resultado_pandas = (
        df_pandas[df_pandas['edad'] > 28]
        .groupby('ciudad')
        .agg({
            'salario': 'mean',
            'edad': 'mean',
            'id': 'count'
        })
        .rename(columns={'id': 'cantidad'})
        .sort_values('salario', ascending=False)
    )
    print(resultado_pandas)
    
    print("\n🐻‍❄️ POLARS:")
    print("""
resultado_polars = (
    df_polars
    .filter(pl.col('edad') > 28)
    .group_by('ciudad')
    .agg([
        pl.col('salario').mean(),
        pl.col('edad').mean(),
        pl.count().alias('cantidad')
    ])
    .sort('salario', descending=True)
)
""")
    resultado_polars = (
        df_polars
        .filter(pl.col('edad') > 28)
        .group_by('ciudad')
        .agg([
            pl.col('salario').mean(),
            pl.col('edad').mean(),
            pl.count().alias('cantidad')
        ])
        .sort('salario', descending=True)
    )
    print(resultado_polars)
    
    # LAZY EVALUATION
    print("\n" + "=" * 70)
    print("6️⃣ LAZY EVALUATION (Optimización Automática)".center(70))
    print("=" * 70)
    
    print("""
🎯 LAZY EVALUATION: Polars optimiza queries antes de ejecutar

EAGER (inmediato):
  df.filter(...).select(...).group_by(...)
  → Ejecuta cada paso inmediatamente

LAZY (perezoso):
  df.lazy().filter(...).select(...).group_by(...).collect()
  → Optimiza todo el pipeline antes de ejecutar
  → Elimina operaciones innecesarias
  → Paraleliza cuando posible
""")
    
    # Crear lazy frame
    print("\n📝 Ejemplo de Lazy Evaluation:")
    
    lazy_query = (
        df_polars.lazy()
        .filter(pl.col('edad') > 25)
        .select(['nombre', 'ciudad', 'salario'])
        .filter(pl.col('salario') > 40000)
        .sort('salario', descending=True)
    )
    
    print("\n1. Query definida (aún no ejecutada):")
    print("   lazy_query = df_polars.lazy().filter(...).select(...)")
    
    print("\n2. Ver plan de ejecución optimizado:")
    print(lazy_query.explain())
    
    print("\n3. Ejecutar con .collect():")
    resultado_lazy = lazy_query.collect()
    print(resultado_lazy)
    
    # INTEROPERABILIDAD
    print("\n" + "=" * 70)
    print("7️⃣ INTEROPERABILIDAD: POLARS ↔ PANDAS".center(70))
    print("=" * 70)
    
    print("\n🔄 Convertir Polars → Pandas:")
    df_to_pandas = df_polars.to_pandas()
    print(f"   Tipo: {type(df_to_pandas)}")
    print(df_to_pandas.head())
    
    print("\n🔄 Convertir Pandas → Polars:")
    df_to_polars = pl.from_pandas(df_pandas)
    print(f"   Tipo: {type(df_to_polars)}")
    print(df_to_polars)
    
    print("\n💡 ESTRATEGIA COMÚN:")
    print("   1. ETL/Transformación pesada → Polars (rápido)")
    print("   2. Convertir a Pandas → .to_pandas()")
    print("   3. Análisis/Visualización → Pandas + Matplotlib/Seaborn")
    
    # BENCHMARK SIMPLE
    print("\n" + "=" * 70)
    print("8️⃣ BENCHMARK: POLARS vs PANDAS".center(70))
    print("=" * 70)
    
    import time
    
    # Crear dataset más grande
    n = 100_000
    df_large_pandas = pd.DataFrame({
        'id': range(n),
        'valor': np.random.randn(n),
        'categoria': np.random.choice(['A', 'B', 'C', 'D'], n),
        'fecha': pd.date_range('2023-01-01', periods=n, freq='1min')
    })
    
    df_large_polars = pl.from_pandas(df_large_pandas)
    
    print(f"\n📊 Dataset de prueba: {n:,} filas")
    
    # Operación: filtrar, agrupar, agregar
    
    # Pandas
    start = time.time()
    result_pandas = (
        df_large_pandas[df_large_pandas['valor'] > 0]
        .groupby('categoria')
        .agg({
            'valor': ['mean', 'sum', 'count'],
            'id': 'max'
        })
    )
    tiempo_pandas = time.time() - start
    
    # Polars (eager)
    start = time.time()
    result_polars = (
        df_large_polars
        .filter(pl.col('valor') > 0)
        .group_by('categoria')
        .agg([
            pl.col('valor').mean().alias('valor_mean'),
            pl.col('valor').sum().alias('valor_sum'),
            pl.col('valor').count().alias('valor_count'),
            pl.col('id').max().alias('id_max')
        ])
    )
    tiempo_polars = time.time() - start
    
    # Polars (lazy)
    start = time.time()
    result_polars_lazy = (
        df_large_polars
        .lazy()
        .filter(pl.col('valor') > 0)
        .group_by('categoria')
        .agg([
            pl.col('valor').mean().alias('valor_mean'),
            pl.col('valor').sum().alias('valor_sum'),
            pl.col('valor').count().alias('valor_count'),
            pl.col('id').max().alias('id_max')
        ])
        .collect()
    )
    tiempo_polars_lazy = time.time() - start
    
    print(f"\n⏱️ RESULTADOS:")
    print(f"  Pandas:         {tiempo_pandas:.4f}s")
    print(f"  Polars (eager): {tiempo_polars:.4f}s  ({tiempo_pandas/tiempo_polars:.1f}x más rápido)")
    print(f"  Polars (lazy):  {tiempo_polars_lazy:.4f}s  ({tiempo_pandas/tiempo_polars_lazy:.1f}x más rápido)")
    
    print(f"\n🏆 Polars es ~{tiempo_pandas/tiempo_polars_lazy:.1f}x más rápido que Pandas en esta operación")

else:
    print("\n💡 Instala Polars para probar los ejemplos:")
    print("   pip install polars")

# RESUMEN POLARS
print("\n" + "=" * 70)
print("RESUMEN: POLARS".center(70))
print("=" * 70)

print("""
✅ VENTAJAS DE POLARS:
  • 5-10x más rápido que Pandas
  • Mejor uso de memoria
  • API expresiva y clara
  • Lazy evaluation (optimización automática)
  • Paralelización sin configuración
  • Excelente con archivos Parquet

⚠️ CONSIDERACIONES:
  • Ecosistema más pequeño que Pandas
  • Menos integración con otras librerías
  • Curva de aprendizaje si vienes de Pandas
  • Algunas operaciones aún no implementadas

🎯 CUÁNDO USAR:
  ✓ Datasets grandes (>1GB)
  ✓ Pipelines ETL en producción
  ✓ Necesitas máxima performance
  ✓ Trabajas con Parquet
  ✓ Transformaciones complejas

🎯 MANTENER PANDAS PARA:
  ✓ Análisis exploratorio rápido
  ✓ Integración con scikit-learn, statsmodels
  ✓ Datasets pequeños
  ✓ Equipo sin experiencia en Polars

💡 MEJOR ESTRATEGIA: Domina Pandas primero, luego aprende Polars
   para casos de uso específicos donde necesites performance.
""")
````

````python
print("=" * 70)
print("STREAMLIT: APPS DE DATOS SIN FRONTEND".center(70))
print("=" * 70)

print("""
🚀 ¿QUÉ ES STREAMLIT?

Framework para convertir scripts Python en aplicaciones web interactivas.

CARACTERÍSTICAS:
  • SCRIPT → APP (sin HTML/CSS/JavaScript)
  • WIDGETS interactivos (sliders, botones, selectores)
  • RE-EJECUTA automáticamente al cambiar inputs
  • DEPLOY fácil (Streamlit Cloud gratuito)
  • INTEGRACIÓN con Plotly, Matplotlib, Pandas

FILOSOFÍA:
  "Si puedes escribir un script Python, puedes hacer una app web"

─────────────────────────────────────────────────────────────

🎯 ¿PARA QUÉ USAR STREAMLIT?

CASOS DE USO PERFECTOS:
  ✓ Prototipos rápidos de ML
  ✓ Dashboards internos de equipo
  ✓ Herramientas de análisis ad-hoc
  ✓ Demos de proyectos
  ✓ POCs para stakeholders
  ✓ Exploración interactiva de datos

NO USAR PARA:
  ✗ Apps de producción enterprise críticas
  ✗ Apps con muchos usuarios simultáneos (>1000)
  ✗ UI/UX muy customizada
  ✗ Apps que requieren control fino del estado

─────────────────────────────────────────────────────────────

🏗️ ARQUITECTURA DE STREAMLIT

1. Escribes script Python normal
2. Agregas comandos st.* (streamlit)
3. Streamlit convierte a app web
4. Cada interacción re-ejecuta el script

EJEMPLO SIMPLE:

```python
import streamlit as st
import pandas as pd

# Título
st.title("Mi Primera App")

# Input
nombre = st.text_input("Tu nombre:")

# Output
if nombre:
    st.write(f"¡Hola, {nombre}!")
    
# Datos
df = pd.DataFrame({'x': [1,2,3], 'y': [4,5,6]})
st.dataframe(df)
```

Ejecutar:
  streamlit run mi_app.py

─────────────────────────────────────────────────────────────

📦 COMPONENTES PRINCIPALES

TEXTO Y MARKDOWN:
  st.title(), st.header(), st.subheader()
  st.text(), st.markdown(), st.latex()
  st.write()  # Automático según tipo

DATOS:
  st.dataframe()  # DataFrame interactivo
  st.table()      # Tabla estática
  st.metric()     # KPI con delta
  st.json()       # JSON formateado

GRÁFICOS:
  st.plotly_chart()    # Plotly
  st.pyplot()          # Matplotlib
  st.altair_chart()    # Altair
  st.line_chart()      # Built-in simple

INPUT WIDGETS:
  st.button()
  st.slider()
  st.selectbox()
  st.multiselect()
  st.text_input()
  st.number_input()
  st.date_input()
  st.file_uploader()

LAYOUT:
  st.sidebar      # Barra lateral
  st.columns()    # Columnas
  st.expander()   # Contenido colapsable
  st.tabs()       # Pestañas
  st.container()  # Agrupar elementos

ESTADO Y CACHE:
  @st.cache_data   # Cachear datos
  st.session_state # Estado entre reruns

─────────────────────────────────────────────────────────────

💻 INSTALACIÓN:

pip install streamlit

Verificar:
  streamlit hello

Ejecutar app:
  streamlit run tu_app.py
""")

print("\n" + "=" * 70)
print("EJEMPLO: APP COMPLETA DE STREAMLIT".center(70))
print("=" * 70)

print("""
📝 A continuación, crearemos una app completa de Streamlit.
   Como no podemos ejecutar Streamlit en este notebook,
   generaremos el código que puedes guardar y ejecutar.

🎯 LA APP INCLUIRÁ:
  • Título y descripción
  • Sidebar con controles
  • Carga de datos
  • Filtros interactivos
  • Visualizaciones con Plotly
  • Métricas y KPIs
  • Descarga de datos
""")

# Código de la app Streamlit
streamlit_app_code = '''
"""
Dashboard de Análisis de Ventas
App creada con Streamlit
"""

import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime, timedelta

# ============================================================
# CONFIGURACIÓN DE PÁGINA
# ============================================================

st.set_page_config(
    page_title="Dashboard de Ventas",
    page_icon="📊",
    layout="wide",
    initial_sidebar_state="expanded"
)

# ============================================================
# FUNCIONES AUXILIARES
# ============================================================

@st.cache_data
def generar_datos_ventas(n=1000):
    """Genera datos de ventas aleatorios"""
    np.random.seed(42)
    
    fechas = pd.date_range(
        start='2023-01-01',
        end='2024-12-31',
        freq='h'
    )[:n]
    
    df = pd.DataFrame({
        'fecha': fechas,
        'producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam'], n),
        'categoria': np.random.choice(['Electrónica', 'Accesorios'], n),
        'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], n),
        'vendedor': np.random.choice(['Ana', 'Bruno', 'Carlos', 'Diana'], n),
        'cantidad': np.random.randint(1, 10, n),
        'precio_unitario': np.random.uniform(20, 1000, n)
    })
    
    df['total'] = df['cantidad'] * df['precio_unitario']
    df['año'] = df['fecha'].dt.year
    df['mes'] = df['fecha'].dt.month
    df['dia_semana'] = df['fecha'].dt.day_name()
    
    return df

# ============================================================
# SIDEBAR
# ============================================================

st.sidebar.header("⚙️ Configuración")

# Selector de fecha
fecha_inicio = st.sidebar.date_input(
    "Fecha inicio",
    value=pd.to_datetime('2023-01-01')
)

fecha_fin = st.sidebar.date_input(
    "Fecha fin",
    value=pd.to_datetime('2024-12-31')
)

# Filtros
st.sidebar.subheader("🔍 Filtros")

# ============================================================
# CARGAR DATOS
# ============================================================

df = generar_datos_ventas()

# Aplicar filtro de fechas
df_filtrado = df[
    (df['fecha'].dt.date >= fecha_inicio) &
    (df['fecha'].dt.date <= fecha_fin)
].copy()

# Filtros de sidebar
regiones = st.sidebar.multiselect(
    "Regiones",
    options=df_filtrado['region'].unique(),
    default=df_filtrado['region'].unique()
)

productos = st.sidebar.multiselect(
    "Productos",
    options=df_filtrado['producto'].unique(),
    default=df_filtrado['producto'].unique()
)

# Aplicar filtros
df_filtrado = df_filtrado[
    (df_filtrado['region'].isin(regiones)) &
    (df_filtrado['producto'].isin(productos))
]

# ============================================================
# HEADER
# ============================================================

st.title("📊 Dashboard de Análisis de Ventas")
st.markdown("---")

# Mostrar info de filtros
col1, col2, col3 = st.columns(3)
with col1:
    st.info(f"📅 Período: {fecha_inicio} - {fecha_fin}")
with col2:
    st.info(f"📍 Regiones: {len(regiones)}")
with col3:
    st.info(f"📦 Productos: {len(productos)}")

st.markdown("---")

# ============================================================
# MÉTRICAS PRINCIPALES (KPIs)
# ============================================================

st.subheader("📈 Métricas Principales")

# Calcular métricas
total_ventas = df_filtrado['total'].sum()
total_transacciones = len(df_filtrado)
ticket_promedio = df_filtrado['total'].mean()
total_unidades = df_filtrado['cantidad'].sum()

# Calcular cambio vs mes anterior (simulado)
cambio_ventas = np.random.uniform(-10, 20)
cambio_transacciones = np.random.uniform(-5, 15)
cambio_ticket = np.random.uniform(-8, 12)
cambio_unidades = np.random.uniform(-3, 18)

# Mostrar métricas en columnas
col1, col2, col3, col4 = st.columns(4)

with col1:
    st.metric(
        label="💰 Total Ventas",
        value=f"${total_ventas:,.0f}",
        delta=f"{cambio_ventas:+.1f}%"
    )

with col2:
    st.metric(
        label="🛒 Transacciones",
        value=f"{total_transacciones:,}",
        delta=f"{cambio_transacciones:+.1f}%"
    )

with col3:
    st.metric(
        label="🎫 Ticket Promedio",
        value=f"${ticket_promedio:,.2f}",
        delta=f"{cambio_ticket:+.1f}%"
    )

with col4:
    st.metric(
        label="📦 Unidades Vendidas",
        value=f"{total_unidades:,}",
        delta=f"{cambio_unidades:+.1f}%"
    )

st.markdown("---")

# ============================================================
# VISUALIZACIONES
# ============================================================

# Tabs para organizar visualizaciones
tab1, tab2, tab3, tab4 = st.tabs(["📈 Tendencias", "📊 Categorías", "🗺️ Regiones", "📋 Datos"])

# --- TAB 1: TENDENCIAS ---
with tab1:
    st.subheader("Evolución de Ventas en el Tiempo")
    
    # Agrupar por fecha
    ventas_tiempo = df_filtrado.groupby(df_filtrado['fecha'].dt.date)['total'].sum().reset_index()
    ventas_tiempo.columns = ['fecha', 'ventas']
    
    fig = px.line(
        ventas_tiempo,
        x='fecha',
        y='ventas',
        title='Ventas Diarias',
        labels={'ventas': 'Ventas ($)', 'fecha': 'Fecha'}
    )
    fig.update_traces(line_color='#1f77b4', line_width=2)
    fig.update_layout(height=400, hovermode='x unified')
    
    st.plotly_chart(fig, use_container_width=True)
    
    # Ventas por día de la semana
    st.subheader("Ventas por Día de la Semana")
    
    ventas_dia = df_filtrado.groupby('dia_semana')['total'].sum().reset_index()
    order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
    ventas_dia['dia_semana'] = pd.Categorical(ventas_dia['dia_semana'], categories=order, ordered=True)
    ventas_dia = ventas_dia.sort_values('dia_semana')
    
    fig = px.bar(
        ventas_dia,
        x='dia_semana',
        y='total',
        title='Distribución Semanal de Ventas',
        labels={'total': 'Ventas ($)', 'dia_semana': 'Día'},
        color='total',
        color_continuous_scale='Blues'
    )
    fig.update_layout(height=400)
    
    st.plotly_chart(fig, use_container_width=True)

# --- TAB 2: CATEGORÍAS ---
with tab2:
    col1, col2 = st.columns(2)
    
    with col1:
        st.subheader("Ventas por Producto")
        
        ventas_producto = df_filtrado.groupby('producto')['total'].sum().reset_index()
        ventas_producto = ventas_producto.sort_values('total', ascending=False)
        
        fig = px.bar(
            ventas_producto,
            x='producto',
            y='total',
            color='producto',
            title='Total de Ventas por Producto'
        )
        fig.update_layout(showlegend=False, height=400)
        
        st.plotly_chart(fig, use_container_width=True)
    
    with col2:
        st.subheader("Distribución por Categoría")
        
        ventas_categoria = df_filtrado.groupby('categoria')['total'].sum().reset_index()
        
        fig = px.pie(
            ventas_categoria,
            values='total',
            names='categoria',
            title='Proporción de Ventas por Categoría',
            hole=0.4
        )
        fig.update_layout(height=400)
        
        st.plotly_chart(fig, use_container_width=True)
    
    # Tabla de productos
    st.subheader("Top 10 Productos")
    
    top_productos = df_filtrado.groupby('producto').agg({
        'total': 'sum',
        'cantidad': 'sum',
        'fecha': 'count'
    }).round(2)
    top_productos.columns = ['Ventas Totales', 'Unidades', 'Transacciones']
    top_productos = top_productos.sort_values('Ventas Totales', ascending=False).head(10)
    
    st.dataframe(top_productos, use_container_width=True)

# --- TAB 3: REGIONES ---
with tab3:
    st.subheader("Análisis por Región")
    
    col1, col2 = st.columns(2)
    
    with col1:
        ventas_region = df_filtrado.groupby('region')['total'].sum().reset_index()
        
        fig = px.bar(
            ventas_region,
            x='region',
            y='total',
            color='region',
            title='Ventas por Región',
            labels={'total': 'Ventas ($)', 'region': 'Región'}
        )
        fig.update_layout(showlegend=False, height=400)
        
        st.plotly_chart(fig, use_container_width=True)
    
    with col2:
        # Vendedores por región
        ventas_vendedor = df_filtrado.groupby(['region', 'vendedor'])['total'].sum().reset_index()
        
        fig = px.bar(
            ventas_vendedor,
            x='region',
            y='total',
            color='vendedor',
            title='Ventas por Región y Vendedor',
            labels={'total': 'Ventas ($)', 'region': 'Región'},
            barmode='group'
        )
        fig.update_layout(height=400)
        
        st.plotly_chart(fig, use_container_width=True)
    
    # Heatmap: producto x región
    st.subheader("Mapa de Calor: Producto x Región")
    
    heatmap_data = df_filtrado.pivot_table(
        values='total',
        index='producto',
        columns='region',
        aggfunc='sum'
    ).fillna(0)
    
    fig = go.Figure(data=go.Heatmap(
        z=heatmap_data.values,
        x=heatmap_data.columns,
        y=heatmap_data.index,
        colorscale='Blues',
        text=np.round(heatmap_data.values, 0),
        texttemplate='$%{text:,.0f}',
        textfont={"size": 10},
        colorbar=dict(title="Ventas ($)")
    ))
    
    fig.update_layout(
        title='Ventas por Producto y Región',
        xaxis_title='Región',
        yaxis_title='Producto',
        height=400
    )
    
    st.plotly_chart(fig, use_container_width=True)

# --- TAB 4: DATOS ---
with tab4:
    st.subheader("📋 Datos Detallados")
    
    # Opciones de visualización
    col1, col2, col3 = st.columns(3)
    
    with col1:
        mostrar_filas = st.number_input(
            "Filas a mostrar",
            min_value=10,
            max_value=len(df_filtrado),
            value=min(100, len(df_filtrado)),
            step=10
        )
    
    with col2:
        ordenar_por = st.selectbox(
            "Ordenar por",
            options=['fecha', 'total', 'cantidad', 'producto']
        )
    
    with col3:
        orden = st.radio(
            "Orden",
            options=['Descendente', 'Ascendente']
        )
    
    # Mostrar datos
    df_mostrar = df_filtrado.sort_values(
        by=ordenar_por,
        ascending=(orden == 'Ascendente')
    ).head(mostrar_filas)
    
    st.dataframe(
        df_mostrar[['fecha', 'producto', 'categoria', 'region', 
                   'vendedor', 'cantidad', 'precio_unitario', 'total']],
        use_container_width=True
    )
    
    # Botón de descarga
    csv = df_filtrado.to_csv(index=False).encode('utf-8')
    
    st.download_button(
        label="📥 Descargar datos filtrados (CSV)",
        data=csv,
        file_name=f'ventas_{datetime.now().strftime("%Y%m%d_%H%M%S")}.csv',
        mime='text/csv'
    )

# ============================================================
# FOOTER
# ============================================================

st.markdown("---")
st.markdown("""
    <div style='text-align: center'>
        <p>Dashboard creado con Streamlit 🚀</p>
        <p>Datos generados aleatoriamente para demostración</p>
    </div>
""", unsafe_allow_html=True)
'''

# Guardar código
print("\n💾 Guardando código de la app Streamlit...")

app_filename = 'dashboard_ventas.py'
with open(app_filename, 'w', encoding='utf-8') as f:
    f.write(streamlit_app_code)

print(f"✅ App guardada en: {app_filename}")

print("\n" + "=" * 70)
print("CÓMO EJECUTAR LA APP".center(70))
print("=" * 70)

print("""
🚀 PASOS PARA EJECUTAR:

1. Asegúrate de tener Streamlit instalado:
   pip install streamlit plotly pandas numpy

2. Ejecuta la app:
   streamlit run dashboard_ventas.py

3. Se abrirá automáticamente en tu navegador
   (típicamente en http://localhost:8501)

4. Interactúa con los controles en el sidebar

5. Para detener la app:
   Ctrl + C en la terminal

─────────────────────────────────────────────────────────────

📝 CARACTERÍSTICAS DE LA APP:

✅ Sidebar con filtros de fecha, región y producto
✅ KPIs principales con indicadores de cambio
✅ Gráficos interactivos con Plotly
✅ Tabs para organizar contenido
✅ Tabla interactiva con datos
✅ Descarga de datos a CSV
✅ Responsive (se adapta al tamaño de pantalla)

─────────────────────────────────────────────────────────────

🚀 DEPLOY EN LA NUBE (GRATIS):

Streamlit Cloud ofrece hosting gratuito:

1. Sube tu código a GitHub
2. Ve a https://streamlit.io/cloud
3. Conecta tu repositorio
4. Deploy automático
5. Obtén URL pública para compartir

Alternativas:
  • Heroku
  • Google Cloud Run
  • AWS Elastic Beanstalk
  • Hugging Face Spaces
""")
````

````python
print("=" * 70)
print("DASH: DASHBOARDS EMPRESARIALES".center(70))
print("=" * 70)

print("""
📊 ¿QUÉ ES DASH?

Framework de Plotly para crear aplicaciones analíticas web.
Más robusto y profesional que Streamlit.

CARACTERÍSTICAS:
  • Basado en React.js (bajo el capó)
  • CALLBACKS para interactividad compleja
  • STATEFUL (manejo explícito de estado)
  • LAYOUTS flexibles con Bootstrap
  • DEPLOYMENT企业级
  • INTEGRACIÓN profunda con Plotly

─────────────────────────────────────────────────────────────

🎯 DASH vs STREAMLIT

┌─────────────────────────────────────────────────────────┐
│                 DASH vs STREAMLIT                       │
├─────────────────────────────────────────────────────────┤
│ FACILIDAD DE USO:                                       │
│   Streamlit: ⭐⭐⭐⭐⭐ (Muy simple)                    │
│   Dash: ⭐⭐⭐ (Más código)                             │
│                                                         │
│ CONTROL:                                                │
│   Streamlit: ⭐⭐⭐ (Convenciones)                      │
│   Dash: ⭐⭐⭐⭐⭐ (Total control)                       │
│                                                         │
│ PERFORMANCE:                                            │
│   Streamlit: ⭐⭐⭐ (Re-ejecuta todo)                   │
│   Dash: ⭐⭐⭐⭐ (Callbacks selectivos)                 │
│                                                         │
│ ESCALABILIDAD:                                          │
│   Streamlit: ⭐⭐⭐ (Hasta ~100 usuarios)               │
│   Dash: ⭐⭐⭐⭐⭐ (Enterprise-grade)                    │
│                                                         │
│ CUSTOMIZACIÓN UI:                                       │
│   Streamlit: ⭐⭐ (Limitada)                            │
│   Dash: ⭐⭐⭐⭐⭐ (CSS/HTML completo)                   │
│                                                         │
│ CURVA APRENDIZAJE:                                      │
│   Streamlit: ⭐ (Muy fácil)                             │
│   Dash: ⭐⭐⭐⭐ (Requiere conceptos web)                │
└─────────────────────────────────────────────────────────┘

CUÁNDO USAR DASH:
  ✓ Apps empresariales de producción
  ✓ Necesitas control fino sobre UI
  ✓ Muchos usuarios simultáneos
  ✓ Callbacks complejos entre componentes
  ✓ Integración con autenticación
  ✓ Custom CSS/branding

CUÁNDO USAR STREAMLIT:
  ✓ Prototipos rápidos
  ✓ Herramientas internas
  ✓ Demos y POCs
  ✓ Menor tráfico
  ✓ Desarrollo más rápido

─────────────────────────────────────────────────────────────

🏗️ ARQUITECTURA DE DASH

1. LAYOUT: Define estructura HTML/componentes
2. CALLBACKS: Define interactividad
3. RUN SERVER: Ejecuta aplicación

COMPONENTES:
  • dash_core_components (dcc): Gráficos, dropdowns, sliders
  • dash_html_components (html): Div, H1, P, etc.
  • dash_bootstrap_components: Componentes Bootstrap

CALLBACKS:
  • Decorador @callback
  • Input: dispara callback
  • Output: what se actualiza
  • State: valor actual sin disparar

EJEMPLO SIMPLE:

```python
from dash import Dash, dcc, html, callback, Input, Output
import plotly.express as px

app = Dash(__name__)

app.layout = html.Div([
    html.H1("Mi Dashboard"),
    dcc.Dropdown(
        id='dropdown',
        options=[{'label': i, 'value': i} for i in ['A', 'B', 'C']],
        value='A'
    ),
    dcc.Graph(id='grafico')
])

@callback(
    Output('grafico', 'figure'),
    Input('dropdown', 'value')
)
def update_graph(valor):
    df = px.data.iris()
    fig = px.scatter(df, x='sepal_width', y='sepal_length')
    return fig

if __name__ == '__main__':
    app.run(debug=True)
```

─────────────────────────────────────────────────────────────

📦 INSTALACIÓN:

pip install dash
pip install dash-bootstrap-components

Ejecutar:
  python tu_app_dash.py
""")

print("\n" + "=" * 70)
print("EJEMPLO: APP DASH COMPLETA".center(70))
print("=" * 70)

dash_app_code = '''
"""
Dashboard Empresarial con Dash
Dashboard profesional para análisis de ventas
"""

from dash import Dash, dcc, html, callback, Input, Output, State
import dash_bootstrap_components as dbc
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime, timedelta

# ============================================================
# GENERAR DATOS
# ============================================================

def generar_datos():
    np.random.seed(42)
    fechas = pd.date_range(start='2023-01-01', end='2024-12-31', freq='D')
    
    df = pd.DataFrame({
        'fecha': fechas,
        'ventas': np.random.randint(1000, 5000, len(fechas)) + 
                 np.cumsum(np.random.randn(len(fechas)) * 50),
        'categoria': np.random.choice(['Electrónica', 'Ropa', 'Alimentos'], len(fechas)),
        'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], len(fechas))
    })
    
    return df

df = generar_datos()

# ============================================================
# INICIALIZAR APP
# ============================================================

app = Dash(
    __name__,
    external_stylesheets=[dbc.themes.BOOTSTRAP],
    suppress_callback_exceptions=True
)

app.title = "Dashboard de Ventas"

# ============================================================
# LAYOUT
# ============================================================

# Navbar
navbar = dbc.Navbar(
    dbc.Container([
        dbc.Row([
            dbc.Col(html.H3("📊 Dashboard de Ventas", className="text-white mb-0"))
        ]),
    ]),
    color="primary",
    dark=True,
    className="mb-4"
)

# Controles
controles = dbc.Card([
    dbc.CardBody([
        html.H5("Filtros", className="card-title"),
        html.Hr(),
        
        # Selector de fecha
        html.Label("Rango de Fechas:"),
        dcc.DatePickerRange(
            id='date-picker',
            start_date=df['fecha'].min(),
            end_date=df['fecha'].max(),
            display_format='DD/MM/YYYY',
            className="mb-3"
        ),
        
        html.Br(),
        html.Br(),
        
        # Selector de región
        html.Label("Región:"),
        dcc.Dropdown(
            id='region-dropdown',
            options=[{'label': 'Todas', 'value': 'ALL'}] +
                   [{'label': r, 'value': r} for r in df['region'].unique()],
            value='ALL',
            className="mb-3"
        ),
        
        # Selector de categoría
        html.Label("Categoría:"),
        dcc.Dropdown(
            id='categoria-dropdown',
            options=[{'label': 'Todas', 'value': 'ALL'}] +
                   [{'label': c, 'value': c} for c in df['categoria'].unique()],
            value='ALL',
            className="mb-3"
        ),
        
        html.Hr(),
        
        # Botón de actualizar
        dbc.Button(
            "🔄 Actualizar Dashboard",
            id='actualizar-btn',
            color="primary",
            className="w-100"
        )
    ])
])

# Métricas (KPIs)
metricas = dbc.Row([
    dbc.Col([
        dbc.Card([
            dbc.CardBody([
                html.H4("💰", className="text-center"),
                html.H5("Total Ventas", className="text-center text-muted"),
                html.H3(id="kpi-ventas", className="text-center")
            ])
        ])
    ], width=3),
    
    dbc.Col([
        dbc.Card([
            dbc.CardBody([
                html.H4("📈", className="text-center"),
                html.H5("Promedio Diario", className="text-center text-muted"),
                html.H3(id="kpi-promedio", className="text-center")
            ])
        ])
    ], width=3),
    
    dbc.Col([
        dbc.Card([
            dbc.CardBody([
                html.H4("📊", className="text-center"),
                html.H5("Mejor Día", className="text-center text-muted"),
                html.H3(id="kpi-mejor", className="text-center")
            ])
        ])
    ], width=3),
    
    dbc.Col([
        dbc.Card([
            dbc.CardBody([
                html.H4("🎯", className="text-center"),
                html.H5("Total Días", className="text-center text-muted"),
                html.H3(id="kpi-dias", className="text-center")
            ])
        ])
    ], width=3),
], className="mb-4")

# Gráficos
graficos = dbc.Row([
    dbc.Col([
        dbc.Card([
            dbc.CardBody([
                html.H5("Evolución de Ventas", className="card-title"),
                dcc.Graph(id='grafico-tiempo')
            ])
        ])
    ], width=8),
    
    dbc.Col([
        dbc.Card([
            dbc.CardBody([
                html.H5("Distribución por Categoría", className="card-title"),
                dcc.Graph(id='grafico-categoria')
            ])
        ])
    ], width=4),
], className="mb-4")

# Tabla
tabla = dbc.Card([
    dbc.CardBody([
        html.H5("Datos Detallados", className="card-title"),
        html.Div(id='tabla-datos')
    ])
])

# Layout principal
app.layout = html.Div([
    navbar,
    dbc.Container([
        dbc.Row([
            dbc.Col(controles, width=3),
            dbc.Col([
                metricas,
                graficos,
                tabla
            ], width=9)
        ])
    ], fluid=True)
])

# ============================================================
# CALLBACKS
# ============================================================

@callback(
    [Output('kpi-ventas', 'children'),
     Output('kpi-promedio', 'children'),
     Output('kpi-mejor', 'children'),
     Output('kpi-dias', 'children'),
     Output('grafico-tiempo', 'figure'),
     Output('grafico-categoria', 'figure'),
     Output('tabla-datos', 'children')],
    [Input('actualizar-btn', 'n_clicks')],
    [State('date-picker', 'start_date'),
     State('date-picker', 'end_date'),
     State('region-dropdown', 'value'),
     State('categoria-dropdown', 'value')]
)
def update_dashboard(n_clicks, start_date, end_date, region, categoria):
    """Actualiza todo el dashboard según filtros"""
    
    # Filtrar datos
    df_filtrado = df.copy()
    
    if start_date and end_date:
        df_filtrado = df_filtrado[
            (df_filtrado['fecha'] >= start_date) &
            (df_filtrado['fecha'] <= end_date)
        ]
    
    if region != 'ALL':
        df_filtrado = df_filtrado[df_filtrado['region'] == region]
    
    if categoria != 'ALL':
        df_filtrado = df_filtrado[df_filtrado['categoria'] == categoria]
    
    # Calcular KPIs
    total_ventas = df_filtrado['ventas'].sum()
    promedio_diario = df_filtrado['ventas'].mean()
    mejor_dia = df_filtrado['ventas'].max()
    total_dias = len(df_filtrado)
    
    # Gráfico de tiempo
    fig_tiempo = px.line(
        df_filtrado,
        x='fecha',
        y='ventas',
        title='Evolución Temporal de Ventas'
    )
    fig_tiempo.update_layout(
        xaxis_title="Fecha",
        yaxis_title="Ventas ($)",
        height=300
    )
    
    # Gráfico de categoría
    ventas_cat = df_filtrado.groupby('categoria')['ventas'].sum().reset_index()
    fig_categoria = px.pie(
        ventas_cat,
        values='ventas',
        names='categoria',
        title='Distribución por Categoría',
        hole=0.4
    )
    fig_categoria.update_layout(height=300)
    
    # Tabla
    df_tabla = df_filtrado.tail(10)[['fecha', 'categoria', 'region', 'ventas']]
    df_tabla['fecha'] = df_tabla['fecha'].dt.strftime('%Y-%m-%d')
    df_tabla['ventas'] = df_tabla['ventas'].apply(lambda x: f"${x:,.2f}")
    
    tabla_component = dbc.Table.from_dataframe(
        df_tabla,
        striped=True,
        bordered=True,
        hover=True,
        responsive=True,
        size='sm'
    )
    
    return (
        f"${total_ventas:,.0f}",
        f"${promedio_diario:,.0f}",
        f"${mejor_dia:,.0f}",
        f"{total_dias}",
        fig_tiempo,
        fig_categoria,
        tabla_component
    )

# ============================================================
# RUN SERVER
# ============================================================

if __name__ == '__main__':
    app.run(debug=True, port=8050)
'''

# Guardar código
print("\n💾 Guardando código de la app Dash...")

dash_filename = 'dashboard_dash.py'
with open(dash_filename, 'w', encoding='utf-8') as f:
    f.write(dash_app_code)

print(f"✅ App Dash guardada en: {dash_filename}")

print("\n" + "=" * 70)
print("CÓMO EJECUTAR LA APP DASH".center(70))
print("=" * 70)

print("""
🚀 PASOS PARA EJECUTAR:

1. Instalar dependencias:
   pip install dash dash-bootstrap-components plotly pandas numpy

2. Ejecutar la app:
   python dashboard_dash.py

3. Abrir navegador en:
   http://localhost:8050

4. Interactuar con filtros y botón "Actualizar Dashboard"

5. Para detener:
   Ctrl + C en la terminal

─────────────────────────────────────────────────────────────

🎨 CARACTERÍSTICAS DE LA APP DASH:

✅ Navbar profesional con branding
✅ Sidebar con controles
✅ 4 KPIs principales
✅ Gráficos interactivos (Plotly)
✅ Tabla de datos recientes
✅ Callbacks para actualización selectiva
✅ Bootstrap para diseño responsive
✅ Botón explícito de actualización

💡 DIFERENCIA vs STREAMLIT:
   • Dash NO re-ejecuta todo automáticamente
   • Necesitas definir callbacks explícitos
   • Más control, pero más código
   • Mejor para apps complejas

─────────────────────────────────────────────────────────────

📚 CONCEPTOS AVANZADOS DE DASH:

1. MULTIPLE OUTPUTS:
   Un callback puede actualizar múltiples componentes

2. CHAINED CALLBACKS:
   Output de uno → Input de otro

3. PREVENT INITIAL CALL:
   prevent_initial_call=True para no ejecutar al cargar

4. LONG CALLBACKS:
   Para operaciones que toman tiempo

5. CLIENT-SIDE CALLBACKS:
   Callbacks en JavaScript para máxima velocidad

6. BACKGROUND CALLBACKS:
   Para tareas asíncronas

7. PATTERN-MATCHING CALLBACKS:
   Para componentes dinámicos (MATCH, ALL, ALLSMALLER)
""")

# Nota: conservamos los archivos generados para poder ejecutarlos fuera del notebook
print("\n📁 Archivos generados conservados: dashboard_ventas.py y dashboard_dash.py")
````
